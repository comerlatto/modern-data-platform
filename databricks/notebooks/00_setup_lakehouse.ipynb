# Configuração inicial do Lakehouse

Este notebook prepara a estrutura inicial da prova de conceito do AdventureWorks no Databricks.

A implementação utiliza o catálogo `workspace` e organiza os dados em três schemas seguindo a arquitetura medalhão:

- **Bronze:** preserva os dados ingeridos das fontes com pouca ou nenhuma transformação;
- **Silver:** concentra limpeza, tipagem, padronização e regras de transformação;
- **Gold:** disponibiliza fatos, dimensões e tabelas preparadas para consumo analítico.

Os comandos utilizam `CREATE SCHEMA IF NOT EXISTS`, permitindo que o notebook seja executado novamente sem recriar schemas existentes ou gerar falhas. Esse comportamento contribui para a reprodutibilidade e a idempotência da configuração.

## Criação dos schemas

A configuração abaixo mantém os nomes e as descrições dos schemas centralizados e cria cada camada no catálogo `workspace`.

```python
catalog_name = "workspace"

schemas = {
    "bronze": "Camada de dados brutos ingeridos das fontes",
    "silver": "Camada de dados limpos, tipados e transformados",
    "gold": "Camada analítica com fatos, dimensões e tabelas de consumo",
}

for schema_name, description in schemas.items():
    spark.sql(
        f"""
        CREATE SCHEMA IF NOT EXISTS {catalog_name}.{schema_name}
        COMMENT '{description}'
        """
    )

print("Schemas configurados com sucesso.")
```

## Validação

A consulta a seguir confirma os schemas disponíveis no catálogo após a execução da configuração.

```python
display(spark.sql("SHOW SCHEMAS IN workspace"))
```

## Volume de arquivos de entrada

O volume gerenciado `adventureworks_files` armazena os arquivos Parquet utilizados como entrada da camada Bronze.

Volumes são destinados ao armazenamento de arquivos no Unity Catalog. As tabelas Delta criadas a partir desses arquivos serão registradas separadamente no schema `workspace.bronze`.

```python
spark.sql(
    """
    CREATE VOLUME IF NOT EXISTS workspace.bronze.adventureworks_files
    COMMENT 'Arquivos Parquet de entrada do AdventureWorks'
    """
)

print("Volume de entrada configurado com sucesso.")
```

```python
display(spark.sql("SHOW VOLUMES IN workspace.bronze"))
```
